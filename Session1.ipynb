```python
for make in df['make'].unique():
    wheel_base=df[df['make']==make]['wheel-base']
    # print(make,wheel_base)
    q1=np.percentile(wheel_base,25)
    q3=np.percentile(wheel_base,75)
    iqr=q3-q1
    uw=q3+1.5*iqr
    inx=wheel_base[wheel_base>uw].index
    if len(inx)!=0:
        df.drop(inx,inplace=True)
    
```

```python
for make in df['make'].unique():
    wheel_base=df[df['make']==make]['wheel-base']
    # print(make,wheel_base)
    q1=np.percentile(wheel_base,25)
    q3=np.percentile(wheel_base,75)
    iqr=q3-q1
    lw=q1-1.5*iqr
    inx=wheel_base[wheel_base<lw].index
    if len(inx)!=0:
        df.drop(inx,inplace=True)
```
